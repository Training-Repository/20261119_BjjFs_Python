````python
import os, subprocess, sys, textwrap

def write_package(root, package, files):
    """Save {file name: source text} as a package folder under root."""
    folder = os.path.join(root, package)
    os.makedirs(folder, exist_ok=True)
    for name, source in files.items():
        with open(os.path.join(folder, name), "w", encoding="utf-8") as f:
            f.write(textwrap.dedent(source).lstrip())

def forget(package):
    """Drop a package and its modules from the import cache."""
    for name in list(sys.modules):
        if name == package or name.startswith(package + "."):
            del sys.modules[name]

def run_script(path):
    """Run a .py file as a script and return what it printed."""
    result = subprocess.run([sys.executable, path], capture_output=True, text=True, encoding="utf-8")
    return (result.stdout + result.stderr).strip()

kyc_requests = [
    # (LAN, PAN, mobile, loan amount ₹)
    ("BFL791", "ABCDE1234F", "9876543210", 250000),
    ("BFL792", "abcde1234f", "+91 98220-12345", 80000.5),
    ("BFL793", "PQRSX9876K", "091-7012345678", 1500000),
]
print(len(kyc_requests), "KYC requests; helpers ready: write_package, forget, run_script")
````
