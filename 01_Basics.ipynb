### Control Structures

```python
val = 50

if val < 50:
    print("Less")
elif val > 50:
    print("More")
    print("Testing")
else:
    print("Equal")
```

```python
# for ( init; condition; increment )
# for i in range()

lst = [1, 2, 3, 4, 5]

srch = 7

for i in lst:
    if i == srch:
        print("Found")
        break
else:
    print("Not Found")
```

```python
val = 5

match val:
    case 1:
        print("One")
    case 2:
        print("Two")
    case 3:
        print("Three")
    case 4:
        print("Four")
    case _:
        print("Default")
```
