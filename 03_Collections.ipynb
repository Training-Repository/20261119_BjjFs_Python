```python
# LISTS
lst = [1, 2, 3, 4, 5]

def Foo(data):
    data.remove(3)
    print("Inside Foo: ", data)

def Bar():
    print("Inside Bar: ", lst)
    Foo(lst[:])
    print("Inside Bar: ", lst)

Bar()
```
