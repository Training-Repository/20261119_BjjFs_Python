```python
def Outer():
    s1 = "Outer string"
    print("Outer --> ", s1)
    vars = globals()
    vars["s1"] = "Modified String"

    print(f"{locals().keys() = }")
    print(f"{globals().keys() = }")

s1 = "Global String"
Outer()
print("Global --> ", s1)
```

```python
# LEGB --  Local, External, Global, Built-in

def Outer():
    s1 = "Outer string"
    def Inner():
        global s1
        # s1 = "Inner String"
        print("Inner --> ", s1)
    Inner()
    print("Outer --> ", s1)

s1 = "Global String"
Outer()
print("Global --> ", s1)
```
